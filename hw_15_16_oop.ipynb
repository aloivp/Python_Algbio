# 1. Design Parking System (LeetCode: 1603)
 https://leetcode.com/problems/design-parking-system/

```python
class ParkingSystem:

    def __init__(self, big: int, medium: int, small: int):
        self.spaces = [big, medium, small]
        

    def addCar(self, carType: int) -> bool:
        i = carType - 1

        if self.spaces[i] == 0:
            return False

        self.spaces[i] -= 1
        return True

        


# Your ParkingSystem object will be instantiated and called as such:
# obj = ParkingSystem(big, medium, small)
# param_1 = obj.addCar(carType)
```

# 2. Simple Bank System (LeetCode: 2043)
 https://leetcode.com/problems/simple-bank-system/description/

```python
class Bank:

    def __init__(self, balance: list[int]):
        self.balance = balance
        self.n = len(balance)


    def transfer(self, account1: int, account2: int, money: int) -> bool:
        if not (1 <= account1 <= self.n):
            return False
        if not (1 <= account2 <= self.n):
            return False
        if self.balance[account1 - 1] < money:
            return False

        self.balance[account1 - 1] -= money
        self.balance[account2 - 1] += money
        return True
        

    def deposit(self, account: int, money: int) -> bool:
        if not (1 <= account <= self.n):
            return False

        self.balance[account - 1] += money
        return True
        

    def withdraw(self, account: int, money: int) -> bool:
        if not (1 <= account <= self.n):
            return False
        if self.balance[account - 1] < money:
            return False

        self.balance[account - 1] -= money
        return True
        


# Your Bank object will be instantiated and called as such:
# obj = Bank(balance)
# param_1 = obj.transfer(account1,account2,money)
# param_2 = obj.deposit(account,money)
# param_3 = obj.withdraw(account,money)
```

# 3. Придумайте свой класс (можно найти в интернете), который реализовывал бы какой-то прикладной аспект в вашей профессии (минимум два метода в классе).

```python
class EpidemicReport:
    def __init__(self, region: str, period: str, population: int, cases: int = 0):
        self.region = region
        self.period = period
        self.population = population
        self.cases = cases

    def add_cases(self, new_cases: int) -> None:
        """Добавляет зарегистрированные случаи за тот же период"""
        self.cases += new_cases


    def cases_per_10k(self) -> float:
        """Рассчитывает число случаев на 10 000 населения"""
        return self.cases / self.population * 1e4
```

```python
report = EpidemicReport(region="Нижний Новгород", period="Неделя 45", population=1e5, cases=120)
print(report.region)
print(report.cases)
report.add_cases(30)
print(report.cases)
print(report.cases_per_10k())
```

# 4. Описать ещё три дополнительных magic метода (один обязательный: __slot__)

```python
class EpidemicReport:
    __slots__ = ("region", "period", "population", "cases")


    def __init__(self, region: str, period: str, population: int, cases: int = 0):
        self.region = region
        self.period = period
        self.population = population
        self.cases = cases

    def add_cases(self, new_cases: int) -> None:
        """Добавляет зарегистрированные случаи за тот же период"""
        self.cases += new_cases


    def cases_per_10k(self) -> float:
        """Рассчитывает число случаев на 10 000 населения"""
        return self.cases / self.population * 1e4
        

    def __eq__(self, other):
        """Сравнивает данные двух отчетов во избежании дублирования информации"""
        if not isinstance(other, EpidemicReport):
            return NotImplemented

        return (self.region, self.period, self.population, self.cases) == (other.region, other.period, other.population, other.cases)
    

    def __repr__(self) -> str:
        """Представление данных отчета в виде строки для удобного отображения"""
        return (f"EpidemicReport(region={self.region!r}, period={self.period!r}, population={self.population!r}, cases={self.cases!r})")
```

```python
report1 = EpidemicReport(region="Нижний Новгород", period="Неделя 45", population=1e5, cases=120)
print(report1.region)
print(report1.cases)
report1.add_cases(30)
print(report1.cases)
print(report1.cases_per_10k())
print(repr(report1))
report2 = EpidemicReport(
    "Нижний Новгород", "Неделя 45", 1e5, 150
)
report3 = EpidemicReport(
    "Нижний Новгород", "Неделя 45", 2e5, 200
)
print(report1 == report2)
print(report1 == report3)
```

# 5. Описать разницу между __str__ и __repr__

оба магических метода нужны для текстового представления объекта  
__str__ - предназначен для пользователя; *print(object)* или *str(obj)*  
__repr__ - предназначен для разработчика, где описана информация об объекте, включающая его тип и значения атрибутов; *repr(obj)*
