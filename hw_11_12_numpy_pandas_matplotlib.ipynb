# NumPy

В эксперименте измерили уровень экспрессии одного гена в пяти образцах. Необходимо проанализировать результаты с помощью NumPy. 

Что нужно сделать:
- Вычислить среднее значение экспрессии
- Найти минимальное и максимальное значения
- Определить индекс образца с максимальной экспрессией
- Создать новый массив, содержащий только значения выше среднего
- Отсортировать исходные значения по возрастанию и вывести результат

```python
import numpy as np
expression = np.array([12, 18, 9, 25, 16])

mean_val = np.mean(expression)
min_val = np.min(expression)
max_val = np.max(expression)
max_index = np.argmax(expression)
above_mean = expression[expression > mean_val]
sorted_expression = np.sort(expression)

print(f"Среднее значение: {mean_val}")
print(f"Минимум: {min_val}")
print(f"Максимум: {max_val}")
print(f"Индекс максимального значения: {max_index}")
print(f"Значения выше среднего: {above_mean}")
print(f"Отсортированный массив: {sorted_expression}")
```

# Pandas

Исследователь измерил экспрессию гена в образцах двух экспериментальных групп. Нужно очистить данные и сравнить группы.  
Что нужно сделать:
- Вывести первые три строки таблицы.
- Найти пропущенные значения в таблице.
- Удалить строки, в которых отсутствует значение экспрессии.
- Отобрать образцы с экспрессией не ниже 18.
- Рассчитать среднюю экспрессию отдельно для групп A и B.  

Ожидаемый результат: очищенная таблица, список образцов с высокой экспрессией и средние значения для каждой группы.

```python
import pandas as pd

df = pd.DataFrame({
    "Образец": ["A1", "A2", "A3", "B1", "B2", "B3"],
    "Группа": ["A", "A", "A", "B", "B", "B"],
    "Экспрессия": [10, 15, None, 20, 25, 18]
})

print("Первые 3 строки:")
print(df.head(3))

print("Пропущенные значения:")
print(df[df["Экспрессия"].isna()])

df_clean = df.dropna()
print("Очищенная таблица:")
print(df_clean)

high_expression = df_clean[df_clean["Экспрессия"] >= 18]
print("Образцы с экспрессией не ниже 18:")
print(high_expression)

group_means = df_clean.groupby("Группа")["Экспрессия"].mean()
print("Средняя экспрессия по группам:")
print(group_means)
```

# Mathplotlib

Постройте график изменения экспрессии гена в зависимости от времени наблюдения  
Что нужно сделать:
- Построить 2D-график зависимости экспрессии от времени.
- Подписать оси: «Время, ч» и «Экспрессия гена».
- Добавить заголовок «Изменение экспрессии гена».
- Отобразить точки измерений маркерами.
- Включить сетку и показать график.

```python
import matplotlib.pyplot as plt

time = [0, 2, 4, 6, 8, 10]
expression = [5, 8, 12, 11, 17, 23]

plt.plot(time, expression, marker='o')
plt.xlabel("Время, ч")
plt.ylabel("Экспрессия гена")
plt.title("Изменение экспрессии гена")
plt.grid(True)
plt.show()
```
